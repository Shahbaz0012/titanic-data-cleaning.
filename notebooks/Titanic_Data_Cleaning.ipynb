# Titanic Survival Prediction — Data Cleaning
This notebook completes the Pandas, missing-data, categorical-encoding, visualization, and CSV-export tasks.

```python
from pathlib import Path
import sys
sys.path.append(str(Path.cwd().parent / 'src'))
from titanic_cleaning import load_data, print_basic_statistics, clean_missing_data, encode_categoricals, save_age_chart

root = Path.cwd().parent
raw_path = root / 'data' / 'raw' / 'titanic.csv'
raw = load_data(raw_path)
raw.head()
```

## 1. Pandas summary statistics

```python
print_basic_statistics(raw)
```

## 2. Missing data: median/mean/mode imputation

```python
cleaned = clean_missing_data(raw)
cleaned.isna().sum()
```

## 3. LabelEncoder and OneHotEncoder

```python
encoded = encode_categoricals(cleaned)
encoded.head()
```

## 4. Age distribution and cleaned CSV

```python
figure_path = root / 'figures' / 'age_distribution.png'
output_path = root / 'data' / 'processed' / 'cleaned_titanic.csv'
save_age_chart(cleaned, figure_path)
output_path.parent.mkdir(parents=True, exist_ok=True)
encoded.to_csv(output_path, index=False)
print(f'Saved: {figure_path}')
print(f'Saved: {output_path}')
```
